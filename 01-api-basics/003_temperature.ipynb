Temperature no longer supported on newer models, replaced by effort (low, medium, high, max, xhigh) which controls how much the model reasons before answering, not how random the wording is.

```python
# Load env variables
from dotenv import load_dotenv

load_dotenv()

# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5"

# Helper functions
def add_user_message(messages, text):
    user_message = ({"role": "user", "content": text})
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = ({"role": "assistant", "content": text})
    messages.append(assistant_message)

def chat(messages, system=None, effort=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    if system:
        params["system"] = system

    if effort:
        params["output_config"] = {"effort": effort}

    message = client.messages.create(**params)
    return message.content[0].text
```

```python
messages = []

add_user_message(messages, "Generate a one sentence movie idea")
answer = chat(messages, effort="low")

answer
```

Important distinction from temperature: this is not a creativity dial. temperature used to control how random/varied the wording was. effort controls how much the model thinks before answering — it trades off thoroughness against speed and token cost. For the "generate a one-sentence movie idea" exercise specifically, low effort is actually the more sensible choice (it's a short, simple task, not one needing deep reasoning) — but it won't make repeated runs more deterministic the way temperature=0.0 did.
